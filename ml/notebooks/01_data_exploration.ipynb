# NERA Risk Intelligence — 01. Exploratory Data Analysis & Provenance
### Target: 8 North Eastern Region (NER) States | SIH 2025

This notebook verifies the **Real-Data Foundation** across the 8 NER states:
1. **NASA Global Landslide Catalog (GLC)**: 442 verified historical disruption incidents (2007–2016)
2. **ERA5 / IMD Daily Gridded Rainfall**: 10-year antecedent rainfall matrix (2007–2016)
3. **Copernicus GLO-30 DEM**: Elevation, derived slope gradient, and terrain ruggedness
4. **HydroRIVERS Asia**: River network proximity & flood exposure

```python
import pandas as pd
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import json
from pathlib import Path

# Check Data Directory
DATA_DIR = Path("../python-service/data")
print(f"Data Directory exists: {DATA_DIR.exists()}")
```

## 1. NASA Global Landslide Catalog (Real Labels)

```python
incidents_path = DATA_DIR / "processed/ner_landslide_incidents.parquet"
if incidents_path.exists():
    df_incidents = pd.read_parquet(incidents_path)
    print(f"Loaded {len(df_incidents)} real historical landslide events in NER.")
    print("\nIncidents by State:")
    print(df_incidents['ner_state'].value_counts())
    
    df_incidents['year'] = pd.to_datetime(df_incidents['event_date']).dt.year
    plt.figure(figsize=(10, 4))
    df_incidents['year'].value_counts().sort_index().plot(kind='bar', color='#10b981')
    plt.title("NASA GLC Historical Disruption Incidents in Northeast India (2007–2016)")
    plt.xlabel("Year")
    plt.ylabel("Confirmed Incident Count")
    plt.grid(axis='y', alpha=0.3)
    plt.show()
```
