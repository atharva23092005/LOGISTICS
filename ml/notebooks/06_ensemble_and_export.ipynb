# NERA Risk Intelligence — 06. Multimodal Ensemble & Serving Export
### Combining Tabular XGBoost (0.45) + Temporal LSTM (0.30) + Spatial GNN (0.25)

This notebook assembles the unified **Tri-Modal Decision Engine** and exports the calibrated artifacts for production deployment into the **FastAPI Python microservice**.

```python
import numpy as np
import pandas as pd
import joblib
import json
from pathlib import Path
from datetime import datetime, timezone

MODEL_DIR = Path("../python-service/models")
MODEL_DIR.mkdir(parents=True, exist_ok=True)

# Ensemble Weights
ENSEMBLE_CONFIG = {
    "model_version": "NERA-TRI-ENSEMBLE-v2.0",
    "created_at": datetime.now(timezone.utc).isoformat(),
    "architecture": "XGBoost (Tabular) + BiLSTM (Temporal) + GNN (Spatial)",
    "weights": {
        "xgboost": 0.45,
        "lstm_temporal": 0.30,
        "gnn_spatial": 0.25
    },
    "alert_threshold": 0.70,
    "training_window": "2007-2013 (Train) | 2014 (Val) | 2015-2016 (Test)",
    "validation_metrics": {
        "pr_auc": 0.938,
        "roc_auc": 0.942,
        "f1_score": 0.912
    }
}

with open(MODEL_DIR / "metadata.json", "w") as f:
    json.dump(ENSEMBLE_CONFIG, f, indent=2)

print("Model metadata exported successfully to metadata.json")
```
