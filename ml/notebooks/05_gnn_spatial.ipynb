# NERA Risk Intelligence — 05. Spatial Graph Neural Network (GNN)
### Modeling Road Network Cascading Disruption & Bottlenecks

Road networks are graphs: Intersections & highway markers are **Nodes**, road sections are **Edges**.
A **Graph Attention Network (GAT)** learns spatial message passing:
- If arterial Pass A is blocked, neighboring feeder corridors absorb stress and exhibit compounding vulnerability.
- Propagates topological risk across isolated mountain corridors.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# Simple Graph Convolution / Attention Layer
class RoadNetworkGNN(nn.Module):
    def __init__(self, in_features=3, hidden_dim=32, out_dim=1):
        super().__init__()
        self.fc1 = nn.Linear(in_features, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, hidden_dim)
        self.out_head = nn.Sequential(
            nn.Linear(hidden_dim, out_dim),
            nn.Sigmoid()
        )

    def forward(self, x, edge_index):
        # Node feature projection
        h = F.relu(self.fc1(x))
        # Message aggregation along road edges
        src, dst = edge_index[0], edge_index[1]
        agg = torch.zeros_like(h)
        agg.index_add_(0, dst, h[src])
        
        h = F.relu(self.fc2(agg))
        return self.out_head(h)

print("Spatial Road Network GNN Architecture Initialized.")
```
